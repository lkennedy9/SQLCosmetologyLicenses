## Question
How big is California's licensed cosmetology workforce, and how do my TikTok Shop affiliate results compare to it?

## Data
| Table | Source file | What it holds |
|---|---|---|
| `licenses_by_county` | interactive_maps.csv | Active licenses by license type, county and fiscal year (FY2016-2023) |
| `annual_reports` | annualreportsdata.csv | Statewide license population, applications, issued and renewed (FY2010-2024) |
| `tiktok_by_type` | tiktok_by_type.csv | My affiliate commission by product type |
| `tiktok_monthly` | tiktok_monthly.csv | My monthly commission by product type (Dec 2025 - Sep 2026) |
| `tiktok_videos` | videos.csv | Commission earned per video |

## Tools
SQL (SQLite) run through Python (pandas), with charts made in seaborn/matplotlib.

## SQL skills shown
CTEs, views, joins, window functions (`LAG`, `RANK`, running totals, moving averages), conditional aggregation, and data cleaning.

## Note on the data
The license data ends in 2023-2024 and my TikTok data starts in Dec 2025, and the two have no shared key such as county. So I use the latest license snapshot as a market-size benchmark instead of joining the tables row by row.


```python
import sqlite3, re
import pandas as pd

conn = sqlite3.connect("cosmetology_tiktok.db")

def snake(col):
    return re.sub(r"[^a-z0-9]+", "_", col.lower()).strip("_")

files = {
    "interactive_maps.csv":  "licenses_by_county",
    "annualreportsdata.csv": "annual_reports",
    "tiktok_by_type.csv":    "tiktok_by_type",
    "tiktok_monthly.csv":    "tiktok_monthly",
    "videos.csv":            "tiktok_videos",
}
for path, table in files.items():
    df = pd.read_csv(path)
    df.columns = [snake(c) for c in df.columns]
    df.to_sql(table, conn, if_exists="replace", index=False)

def run(sql):
    return pd.read_sql(sql, conn)

run("""
SELECT 'licenses_by_county' AS table_name, COUNT(*) AS row_count FROM licenses_by_county
UNION ALL SELECT 'annual_reports', COUNT(*) FROM annual_reports
UNION ALL SELECT 'tiktok_by_type', COUNT(*) FROM tiktok_by_type
UNION ALL SELECT 'tiktok_monthly', COUNT(*) FROM tiktok_monthly
UNION ALL SELECT 'tiktok_videos',  COUNT(*) FROM tiktok_videos
""")
```

```python
conn.executescript("""
DROP VIEW IF EXISTS cosmo_county;
CREATE VIEW cosmo_county AS
SELECT CAST(SUBSTR(fiscal_year, -4) AS INTEGER) AS fiscal_year,
       license_type_name, county, active_licenses
FROM licenses_by_county
WHERE board_long_name = 'Board of Barbering and Cosmetology';

DROP VIEW IF EXISTS cosmo_statewide;
CREATE VIEW cosmo_statewide AS
SELECT CAST(SUBSTR(date, -4) AS INTEGER) AS fiscal_year,
       license_type,
       CAST(REPLACE(license_population,    ',', '') AS INTEGER) AS license_population,
       CAST(REPLACE(applications_received, ',', '') AS INTEGER) AS applications_received,
       CAST(REPLACE(license_issued,        ',', '') AS INTEGER) AS license_issued,
       CAST(REPLACE(license_renewed,       ',', '') AS INTEGER) AS license_renewed
FROM annual_reports
WHERE dca_entity = 'Board of Barbering and Cosmetology';

DROP TABLE IF EXISTS product_license_map;
CREATE TABLE product_license_map (product_type TEXT PRIMARY KEY, license_type TEXT);
INSERT INTO product_license_map VALUES
  ('Clip-in','Cosmetologist'), ('Tape-in','Cosmetologist'),
  ('Accessory','Cosmetologist'), ('Other','Cosmetologist');
""")
```

```python
run("""
SELECT fiscal_year,
       SUM(CASE WHEN license_type_name='Cosmetologist' THEN active_licenses END) AS cosmetologists,
       SUM(CASE WHEN license_type_name='Esthetician'   THEN active_licenses END) AS estheticians,
       SUM(CASE WHEN license_type_name='Manicurist'    THEN active_licenses END) AS manicurists,
       SUM(CASE WHEN license_type_name='Establishment' THEN active_licenses END) AS establishments
FROM cosmo_county
GROUP BY fiscal_year
ORDER BY fiscal_year
""")
```

```python
run("""
WITH yearly AS (
  SELECT fiscal_year, SUM(active_licenses) AS cosmetologists
  FROM cosmo_county
  WHERE license_type_name = 'Cosmetologist'
  GROUP BY fiscal_year
)
SELECT fiscal_year, cosmetologists,
       LAG(cosmetologists) OVER (ORDER BY fiscal_year) AS prior_year,
       ROUND(100.0 * (cosmetologists - LAG(cosmetologists) OVER (ORDER BY fiscal_year))
             / LAG(cosmetologists) OVER (ORDER BY fiscal_year), 2) AS yoy_pct
FROM yearly
ORDER BY fiscal_year
""")
```

```python
run("""
WITH county_totals AS (
  SELECT county, SUM(active_licenses) AS active_licenses
  FROM cosmo_county WHERE fiscal_year = 2023
  GROUP BY county
),
ranked AS (
  SELECT county, active_licenses,
         RANK() OVER (ORDER BY active_licenses DESC) AS county_rank,
         ROUND(100.0 * active_licenses / SUM(active_licenses) OVER (), 1) AS pct_of_state
  FROM county_totals
)
SELECT * FROM ranked WHERE county_rank <= 5 ORDER BY county_rank
""")
```

```python
top5_df = run("""
WITH top5 AS (
  SELECT county FROM cosmo_county
  WHERE fiscal_year = 2023
  GROUP BY county
  ORDER BY SUM(active_licenses) DESC
  LIMIT 5
)
SELECT c.county AS County,
       c.license_type_name AS "License Type Name",
       SUM(c.active_licenses) AS Total_Active_Licenses
FROM cosmo_county c
JOIN top5 t ON t.county = c.county
WHERE c.fiscal_year = 2023
  AND c.license_type_name IN ('Cosmetologist','Esthetician','Manicurist','Establishment')
GROUP BY c.county, c.license_type_name
ORDER BY c.county, Total_Active_Licenses DESC
""")
top5_df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

tt = run("SELECT product_type, commission FROM tiktok_by_type ORDER BY commission DESC")
sns.barplot(data=tt, x="product_type", y="commission", hue="product_type",
            ax=axes[0], palette="viridis", legend=False)
axes[0].set_title("TikTok Shop Commission by Hair Product Type ($)", fontsize=14)
axes[0].set_xlabel("")
axes[0].set_ylabel("Total Commission ($)")

order = top5_df.groupby("County")["Total_Active_Licenses"].sum().sort_values(ascending=False).index
sns.barplot(data=top5_df, x="Total_Active_Licenses", y="County", hue="License Type Name",
            order=order, ax=axes[1], palette="magma")
axes[1].set_title("Top 5 CA Counties with Active Cosmetology Licenses (FY2023)", fontsize=14)
axes[1].set_xlabel("Count of Active Licenses")
axes[1].legend(title="License Type", loc="lower right", frameon=True)

plt.tight_layout()
plt.show()
```

```python
run("""
SELECT fiscal_year, license_population, applications_received, license_issued,
       ROUND(100.0 * license_issued / NULLIF(applications_received, 0), 1) AS issued_per_100_apps,
       license_renewed
FROM cosmo_statewide
WHERE license_type = 'Cosmetologist License' AND fiscal_year >= 2018
ORDER BY fiscal_year
""")
```

```python
run("""
SELECT product_type, commission,
       ROUND(100.0 * commission / SUM(commission) OVER (), 1) AS pct_of_total
FROM tiktok_by_type
ORDER BY commission DESC
""")
```

```python
run("""
WITH monthly AS (
  SELECT month, ROUND(accessory + clip_in + other + tape_in, 2) AS total_commission
  FROM tiktok_monthly
)
SELECT month, total_commission,
       ROUND(100.0 * (total_commission - LAG(total_commission) OVER (ORDER BY month))
             / LAG(total_commission) OVER (ORDER BY month), 1) AS mom_pct,
       ROUND(SUM(total_commission) OVER (ORDER BY month), 2) AS running_total,
       ROUND(AVG(total_commission) OVER (ORDER BY month
             ROWS BETWEEN 2 PRECEDING AND CURRENT ROW), 2) AS moving_avg_3mo
FROM monthly
ORDER BY month
""")
```

```python
run("""
WITH ranked AS (
  SELECT video_id, commission,
         ROW_NUMBER() OVER (ORDER BY commission DESC) AS video_rank
  FROM tiktok_videos
)
SELECT video_rank, video_id, commission,
       ROUND(100.0 * SUM(commission) OVER (ORDER BY video_rank)
             / SUM(commission) OVER (), 1) AS cumulative_pct
FROM ranked
ORDER BY video_rank
LIMIT 10
""")
```

```python
run("""
WITH latest_licenses AS (
  SELECT license_type_name, SUM(active_licenses) AS active_licenses
  FROM cosmo_county
  WHERE fiscal_year = (SELECT MAX(fiscal_year) FROM cosmo_county)
  GROUP BY license_type_name
)
SELECT t.product_type, m.license_type, t.commission, l.active_licenses,
       ROUND(1000.0 * t.commission / l.active_licenses, 2) AS commission_per_1000_licenses
FROM tiktok_by_type t
JOIN product_license_map m ON m.product_type = t.product_type
JOIN latest_licenses l     ON l.license_type_name = m.license_type
ORDER BY t.commission DESC
""")
```

```python
run("""
WITH licenses AS (
  SELECT SUM(CASE WHEN fiscal_year = 2023 THEN active_licenses END) AS cosmetologists_2023,
         SUM(CASE WHEN fiscal_year = 2016 THEN active_licenses END) AS cosmetologists_2016
  FROM cosmo_county
  WHERE license_type_name = 'Cosmetologist'
),
monthly AS (
  SELECT month, accessory + clip_in + other + tape_in AS total FROM tiktok_monthly
),
tiktok AS (
  SELECT ROUND(SUM(total), 2) AS total_commission,
         ROUND(AVG(total), 2) AS avg_monthly,
         ROUND(MAX(total), 2) AS best_month,
         (SELECT month FROM monthly ORDER BY total DESC LIMIT 1) AS best_month_name,
         ROUND((SELECT AVG(total) FROM (SELECT total FROM monthly ORDER BY month DESC LIMIT 3)), 2) AS last_3mo_avg,
         ROUND((SELECT AVG(total) FROM (SELECT total FROM monthly ORDER BY month ASC  LIMIT 3)), 2) AS first_3mo_avg
  FROM monthly
)
SELECT l.cosmetologists_2023,
       ROUND(100.0 * (l.cosmetologists_2023 - l.cosmetologists_2016) / l.cosmetologists_2016, 1) AS license_growth_pct_2016_2023,
       t.total_commission, t.avg_monthly, t.best_month_name, t.best_month,
       ROUND(1000.0 * t.total_commission / l.cosmetologists_2023, 2) AS commission_per_1000_cosmetologists,
       t.first_3mo_avg, t.last_3mo_avg,
       ROUND(100.0 * (t.last_3mo_avg - t.first_3mo_avg) / t.first_3mo_avg, 1) AS tiktok_3mo_change_pct
FROM licenses l
CROSS JOIN tiktok t
""")
```

```python
 import matplotlib.pyplot as plt
import seaborn as sns

yearly = run("""
SELECT fiscal_year, SUM(active_licenses) AS cosmetologists
FROM cosmo_county
WHERE license_type_name = 'Cosmetologist'
GROUP BY fiscal_year
ORDER BY fiscal_year
""")

sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 5))
ax = sns.lineplot(data=yearly, x="fiscal_year", y="cosmetologists",
                  marker="o", linewidth=2.5, color="#6a1b9a")

for _, r in yearly.iterrows():
    ax.annotate(f"{r.cosmetologists:,.0f}", (r.fiscal_year, r.cosmetologists),
                textcoords="offset points", xytext=(0, 10), ha="center", fontsize=9)

ax.set_title("Active Cosmetologist Licenses in California by Fiscal Year", fontsize=14)
ax.set_xlabel("Fiscal Year")
ax.set_ylabel("Active Licenses")
ax.set_ylim(215000, 260000)
plt.tight_layout()
plt.show()
```

## Summary of Findings

**Market (California license data)**
- Active cosmetologist licenses grew 4.5% from FY2016 to FY2023, but dipped 3.4% in FY2023.
- Los Angeles alone holds 24% of the state's cosmetology board licenses. The top 5 counties are Los Angeles, Orange, San Diego, Riverside and San Bernardino.

**My TikTok Shop performance**
- One video drove 72% of my commission, and my top 5 videos drove 96%.
- Clip-in extensions make up 99% of my commission.
- My best month was April 2026 ($644).

- I earned about $14.48 in commission per 1,000 licensed cosmetologists.


**Takeaways**
- My income depends heavily on one video and one product type, so creating more videos and expanding into other hair products would reduce that risk.
-Capitalize on Titkok shop due to a decline in cosmetology licenses, and people are more willing to DIY at home versus go to a salon.

**Limitations**
- The license data ends in 2023-2024, and my TikTok data starts in Dec 2025, so I compared against the latest license snapshot instead of joining the two by date or county.
